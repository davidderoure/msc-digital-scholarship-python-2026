````python
len(sample)
````

## 6. Modules: `module.function(...)`

Python comes with extra collections of functions called **modules**. Think of a module as a toolbox:
before you can use its tools, you bring it in with `import`. Then you call its functions with the module name, a dot, and the function name.

Here's the `statistics` module, used to average the lengths of the first five words:

````python
import statistics

statistics.mean([3, 6, 6, 2, 4])
````

This looks a lot like a method (`sample.split()`), because both use a dot. The difference is what comes **before** the dot:

| Pattern | Example | Read it as |
|---|---|---|
| `function(thing)` | `len(sample)` | "do `len` to `sample`" |
| `thing.method(...)` | `sample.split()` | "take `sample` and `split` it" |
| `module.function(thing)` | `statistics.mean(numbers)` | "use `mean` from the `statistics` module, on `numbers`" |

In the next notebook you'll meet `re.split(pattern, sample)`. That's the **third** pattern:
`split` from the `re` module (short for "**r**egular **e**xpressions"). It does a similar job to `sample.split(...)`, but it's a different tool,
so the text goes **inside** the brackets instead of before the dot.

**Vocabulary you'll meet.** A **module** is a single file of Python code you can `import`. A **package** is a collection of modules
that usually has to be installed first (Colab has most of the ones we need already). People, and AI assistants, often say **library**
for either. So "the NLTK library", "the NLTK package" and "NLTK's modules" all refer to the same kind of thing.

## 7. Doing something to every item: list comprehensions

Often we want to do the same thing to every item in a list. Python has a compact way to say
"**make a new list by doing this to each item of that list**". It's called a **list comprehension**:

````python
[len(w) for w in words]
````

Read it from the middle outwards: "**for each `w` in `words`, give me `len(w)`**", all collected in a new list (the square brackets).
The name `w` is just a temporary label for "the current item". You could call it anything.

Here are the lengths of the first five words, which we averaged by hand in Section 6:

````python
[len(w) for w in words[0:5]]
````

````python
statistics.mean([len(w) for w in words[0:5]])
````

### Keeping only some items
Add `if` at the end to **filter**: keep only the items that pass a test.

````python
[w for w in words if len(w) > 5]
````

````python
[w for w in words if w.startswith("B")]
````

### When you want to *do* something rather than *make* something: `for` loops
A list comprehension **makes a new list**. Sometimes we don't want a list. We want an *action*, such as printing each item on its own line.
For that we use a `for` loop:

````python
for w in words[0:8]:
    print(w, len(w))
````

Two details matter:
- The first line ends with a colon `:`.
- The lines to repeat are **indented** (pushed in by four spaces). Python uses indentation to know what belongs inside the loop.

## 9. Putting it together: a concordance

A **concordance** lists every occurrence of a word, each shown in its surrounding context. Concordances are among the oldest
tools of textual scholarship, and they used to take enormous human effort:
- Alexander Cruden compiled his concordance to the King James Bible (1737) by hand.
- In 1949 the Jesuit scholar Roberto Busa persuaded IBM to help him build a concordance to the works of Thomas Aquinas, using punched cards.
  The *Index Thomisticus* took some thirty years to complete, and is often counted as the beginning of humanities computing.

Let's build one with the tools from this notebook.

### Finding where a word occurs
`enumerate` gives us each item of a list **together with its position**. So this comprehension means
"**the position `i` of every word `w` that is `Bennet`**":

````python
[i for i, w in enumerate(words) if w == "Bennet"]
````

Now we can use slicing to show each one in context: three words before, and three words after.

````python
[words[i - 3:i + 4] for i, w in enumerate(words) if w == "Bennet"]
````

> **Critical check.** One of these is wrong. Which, and why? (Hint: what is `1 - 3`? What does a negative number mean in a slice? See Section 5.)

### A concordance function
Here is a tidier version. We'll write two small **functions** of our own (`def` defines one; `return` says what it gives back):
- `clean` strips punctuation off a word and makes it lower case, so that `"Bennet's,` and `bennet` can be compared fairly.
- `concordance` lines up each occurrence in the middle, with its context to the left and right.

`max(0, i - width)` is the fix for the problem you just found: never start a slice before position 0.

````python
def clean(w):
    return w.strip('.,;:!?"“”‘’()_-').lower()

def concordance(word_list, keyword, width=5):
    return [" ".join(word_list[max(0, i - width):i]).rjust(45) + "   " + w + "   " + " ".join(word_list[i + 1:i + 1 + width])
            for i, w in enumerate(word_list) if clean(w) == keyword]

concordance(words, "bennet")
````
